```python
import json
import pandas as pd
from io import StringIO
import matplotlib.pyplot as plt

# Specify the path to your JSON file
json_file_path = 'C:/Users/nicho/OneDrive/Desktop/Vaccine_Tender/horizon excursion/15 years - Tender cost sensitivity/tender_results.json' 

# Load JSON data from the file
with open(json_file_path, 'r') as file:
    data_list = [json.loads(line) for line in file]

# Create an empty list to store DataFrames
dfs = []

# Iterate through the list of dictionaries and create DataFrames
for entry in data_list:
    tender_cost = entry["tender_cost"]
    obj_value = entry["obj_value"]
    df_entry = pd.DataFrame({'tender_cost': [tender_cost], 'OBJ': [obj_value]})
    dfs.append(df_entry)

# Concatenate the list of DataFrames into a single DataFrame
df = pd.concat(dfs, ignore_index=True)

df = df.sort_values(by='tender_cost')
```

```python
# Plot the DataFrame
plt.plot(df['tender_cost'], df['OBJ'], marker='o', linestyle='-')
plt.xlabel('tender_cost')
plt.ylabel('OBJ')
plt.title('Plot of tender_cost vs. OBJ')
plt.grid(True)
plt.show()
```
