```python
import json

def load_json_file(file_path):
    with open(file_path, 'r') as file:
        return json.load(file)

def analyze_iis_constraints(json_data):
    # Check and print constraints excluding those containing {'type': 'ZeroOne'}
    if 'constraints' in json_data:
        for constraint in json_data['constraints']:
            constraint_str = json.dumps(constraint)
            if '{"type": "ZeroOne"}' not in constraint_str:  # Check the JSON string representation
                print(f"Constraint: {constraint}")
    else:
        print("No constraints part found in JSON data.")
    
if __name__ == "__main__":
    file_path = 'errors.json'  # Path to the JSON file
    json_data = load_json_file(file_path)
    analyze_iis_constraints(json_data)
```
