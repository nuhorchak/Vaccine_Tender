```python
import pandas as pd

# Load the uploaded CSV file
file_path = 'data/pair_demand_capacity_new_demand_base_capacity_pandemic.csv'
data = pd.read_csv(file_path)

# Create an empty dictionary for scenario_pairs
expanded_scenario_pairs = {}

# Iterate over the rows to populate the dictionary
for _, row in data.iterrows():
    pair_idx = row['pair_idx']
    demand_or_capacity = row['type']
    values = eval(row['values'])  # Convert string representation of list to an actual list
    unit = row['unit']
    
    if pair_idx not in expanded_scenario_pairs:
        expanded_scenario_pairs[pair_idx] = {'demand': {}, 'capacity': {}}
    
    if demand_or_capacity == 'antigen':
        for i, value in enumerate(values):
            year = i + 1
            if year not in expanded_scenario_pairs[pair_idx]['demand']:
                expanded_scenario_pairs[pair_idx]['demand'][year] = {}
            expanded_scenario_pairs[pair_idx]['demand'][year][unit] = value
    elif demand_or_capacity == 'manufacturer':
        for i, value in enumerate(values):
            year = i + 1
            if year not in expanded_scenario_pairs[pair_idx]['capacity']:
                expanded_scenario_pairs[pair_idx]['capacity'][year] = {}
            expanded_scenario_pairs[pair_idx]['capacity'][year][unit] = value

expanded_scenario_pairs

```

```python
import json

# Define the file path for the JSON output
json_output_path = 'data/scenario_pairs_new_demand_base_capacity_pandemic.json'

# Write the dictionary to a JSON file
with open(json_output_path, 'w') as json_file:
    json.dump(expanded_scenario_pairs, json_file)

# Provide the file path for download
json_output_path
```

```python
import pandas as pd

# Load the CSV file into a DataFrame
file_path = 'data/pair_demand_capacity_new_demand_base_capacity_pandemic.csv'
df = pd.read_csv(file_path)

# Create the dictionary with scenarios and their probabilities
scenarios_probabilities = df[['pair_idx', 'pair_probability']].drop_duplicates().set_index('pair_idx')['pair_probability'].to_dict()

json_file_path = 'data/scenario_pair_probabilities_new_demand_base_capacity_pandemic.json'

# Write the dictionary to a JSON file
with open(json_file_path, 'w') as json_file:
    json.dump(scenarios_probabilities, json_file, indent=4)

print(f"Scenarios probabilities have been saved to {json_file_path}")
```
