```python
from fitter import Fitter
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
from fitter import get_common_distributions
```

```python
data = pd.read_excel('AR1.xlsx', skiprows = 1, sheet_name = 'Residuals')
```

```python
data
```

```python
dists = get_common_distributions()
dists
```

```python
# Function to bootstrap a single column
def bootstrap_column(column, n=100):
    # Resample with replacement to reach the desired number of samples
    return np.random.choice(column, size=n, replace=True)

# Creating a new DataFrame with bootstrapped columns
bootstrapped_df = pd.DataFrame({col: bootstrap_column(data[col].dropna(), n=100) for col in data.columns})

# Now, bootstrapped_df is your new DataFrame with 100 rows obtained by bootstrapping each column.
bootstrapped_df
```

```python
# Dictionary to store the best fit for each column
# best_fits = {}
fit_results = pd.DataFrame(columns=['Column', 'Best Distribution', 'Parameters', 'Sum Square Error'])


# Iterate over each column in the DataFrame
for column in bootstrapped_df.columns:
    # Use Fitter on the column data
    f = Fitter(bootstrapped_df[column].dropna(), distributions = dists)
    f.fit()
    # Get the best fitting distribution based on sum of square errors
    # best_fits[column] = f.get_best(method = 'sumsquare_error')
    best_fit = f.get_best(method='sumsquare_error')
    
    # Extract the best fitting distribution and its parameters
    best_distribution, params = list(best_fit.items())[0]

    # Extract sum square error for the best fitting distribution
    sse = f.df_errors.loc[best_distribution, 'sumsquare_error']

    # Append results to the DataFrame
    fit_results = fit_results.append({
        'Column': column,
        'Best Distribution': best_distribution,
        'Parameters': params,
        'Sum Square Error': sse
    }, ignore_index=True)
    
#     f.summary()
#     # Access the current figure
#     fig = plt.gcf()  # Get the current figure after calling f.summary()

#     # Dynamically create the file name
#     file_name = f'{column}_distribution_fit_summary.png'

#     # Save the figure with the dynamic file name
#     fig.savefig(file_name)

#     # Close the figure to free memory and avoid overlap in subsequent plots
#     plt.close(fig)

# Print the best fits for each column
best_fits
```
