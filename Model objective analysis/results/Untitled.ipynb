```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load the JSON file
with open('DE_results_T_10_delta_5_scen_4_trial_1_inv_1_cap._1_cap.inc._2_sensitivity_max_profit_min_unvax.json') as f:
    data = json.load(f)

# Extract the unvac_children data
unvac_data = data['unvac_children']

# Convert it into a DataFrame
df_unvac = pd.DataFrame(unvac_data).transpose()

# Melt the DataFrame to have a column for years (1-4) and values
df_unvac_melted = df_unvac.reset_index().melt(id_vars='index', var_name='Year', value_name='Amount')

# Sort the data to align with plotting expectations
df_unvac_melted.sort_values(by=['Year', 'index'], inplace=True)

# Plotting
plt.figure(figsize=(12, 8))
years = df_unvac_melted['Year'].unique()
indices = np.arange(len(df_unvac_melted['index'].unique()))  # Create unique positions for vaccines

# Set the width for each bar
bar_width = 0.2

# Create bar plots for each year
for i, year in enumerate(years):
    year_data = df_unvac_melted[df_unvac_melted['Year'] == year]
    plt.bar(indices + i * bar_width, year_data['Amount'], width=bar_width, label=f'Year {year}')

# Add labels and title
plt.xlabel('Vaccines')
plt.ylabel('Amount of Unvaccinated Children')
plt.title('Unvaccinated Children by Vaccine and Scenario')
plt.xticks(indices + bar_width * (len(years) - 1) / 2, df_unvac_melted['index'].unique(), rotation=90)
plt.legend()

# Display the plot
plt.tight_layout()
plt.savefig('max_profit.png')
plt.show()
```
