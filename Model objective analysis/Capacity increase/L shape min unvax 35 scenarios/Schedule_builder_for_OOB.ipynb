```python
import json
import pandas as pd
```

```python
# Load the provided JSON file to locate the "F" key
file_path = 'DE_L_results_T_10_delta_5_scen_35_trial_1_inv_1_cap._1_cap.inc._1_min_unvax.json'
with open(file_path, 'r') as file:
    data = json.load(file)

# Check if the "F" key exists and extract its content
f_data = data.get("F")
f_data
```

```python
# Prepare the extracted data for scheduling
schedule_data = []

for antigen, values in f_data.items():
    for start, end_dict in values.items():
        for end, value in end_dict.items():
            if value == 1:  # If value in 'END' is 1, create the schedule entry
                schedule_data.append({'Antigen': antigen, 'Starting': int(start), 'Ending': int(end)})

# Convert to a DataFrame and display
schedule_df = pd.DataFrame(schedule_data)
schedule_df.sort_values(by=['Antigen', 'Starting'], inplace=True)
```

```python
schedule_df.sort_values(by=['Starting'], inplace=True)
# schedule_df
```

```python
schedule_df.to_excel("min_missed_dose_fixed_schedule.xlsx")
```
