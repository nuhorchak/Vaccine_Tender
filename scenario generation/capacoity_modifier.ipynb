```python
import pandas as pd
import numpy as np


# Load the CSV file to see its content
file_path = 'data/production_capacity_scenarios.csv'
df = pd.read_csv(file_path)

# Display the first few rows of the DataFrame to understand its structure
df.head()
```

```python
# Function to apply a random +/- 2% modification to the capacity values
def modify_capacity(capacity_list):
    modified_list = [
        value * (1 + np.random.uniform(-0.02, 0.02)) for value in capacity_list
    ]
    return modified_list

# Apply the modification to every entry in the 'capacity' column
df['capacity'] = df['capacity'].apply(lambda x: modify_capacity(eval(x)))

# Display the first few rows of the DataFrame to understand its structure
df.head()
```
