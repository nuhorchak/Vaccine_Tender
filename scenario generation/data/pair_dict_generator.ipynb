```python
import pandas as pd

# Load the uploaded CSV file
file_path = 'pair_demand_capacity.csv'
data = pd.read_csv(file_path)

# Create an empty dictionary for scenario_paris
scenario_paris = {}

# Iterate over the rows to populate the dictionary
for _, row in data.iterrows():
    pair_idx = row['pair_idx']
    demand_or_capacity = row['type']
    values = eval(row['values'])  # Convert string representation of list to an actual list
    
    if pair_idx not in scenario_paris:
        scenario_paris[pair_idx] = {'demand': {}, 'capacity': {}}
    
    if demand_or_capacity == 'antigen':
        scenario_paris[pair_idx]['demand'] = {i+1: values[i] for i in range(len(values))}
    elif demand_or_capacity == 'manufacturer':
        scenario_paris[pair_idx]['capacity'] = {i+1: values[i] for i in range(len(values))}

# Display the resulting dictionary structure
# scenario_paris
```

```python
import json

# Define the file path for the JSON output
json_output_path = 'scenario_paris.json'

# Write the dictionary to a JSON file
with open(json_output_path, 'w') as json_file:
    json.dump(scenario_paris, json_file)

# Provide the file path for download
json_output_path
```
