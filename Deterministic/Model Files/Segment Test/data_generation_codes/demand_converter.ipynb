```python
import pandas as pd
import sys

def transform_excel_to_csv(input_file, output_file):
    """
    Transform Excel file from wide format to long format with demand lists.
    
    Args:
        input_file: Path to input Excel file
        output_file: Path to output CSV file
    """
    # Read the Excel file
    df = pd.read_excel(input_file, sheet_name="Medium_Demand")
    
    # Get antigen names (first column)
    antigens = df.iloc[:, 0].values
    
    # Get year columns (all columns except the first one)
    year_columns = df.columns[1:]
    
    # Create output data
    output_data = []
    
    for antigen in antigens:
        # Get the row for this antigen
        antigen_row = df[df.iloc[:, 0] == antigen]
        
        # Extract demand values for all years
        demand_values = antigen_row.iloc[:, 1:].values.flatten().tolist()
        
        # Format as a list string
        demand_list = '[' + ', '.join(map(str, demand_values)) + ']'
        
        # Add to output data
        output_data.append({
            'antigen': antigen,
            'demand': demand_list,
            'probability': 1,
            'demand_SID': 1
        })
    
    # Create output dataframe
    output_df = pd.DataFrame(output_data)
    
    # Write to CSV
    output_df.to_csv(output_file, index=False)
    print(f"Transformation complete! Output saved to: {output_file}")
    print(f"Processed {len(antigens)} antigens with {len(year_columns)} years of data each")

```

```python
   
input_file = "data/Demand_Scenarios_Base_15YR.xlsx"
output_file = "data/Medium_Demand_Structured_15YR.csv"
    
try:
    transform_excel_to_csv(input_file, output_file)
except Exception as e:
    print(f"Error: {e}")
    sys.exit(1)
```
