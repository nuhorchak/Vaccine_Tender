```python
import json
import plotly.express as px
import pandas as pd
```

```python
with open('Deterministic_results_with_start.json') as json_file:
    F_results = json.load(json_file)

# print(F_results)
```

```python

# List of keywords to search for
keywords_to_search = ["F","Q","X"]

# Initialize a dictionary to store grouped results
grouped_results = {keyword: [] for keyword in keywords_to_search}

# Search for keywords in the JSON keys and group the results
for key, value in F_results.items():
    for keyword in keywords_to_search:
        if keyword in key:
            grouped_results[keyword].append({key: value})

# Print the grouped results
for keyword, results in grouped_results.items():
    print(f'Keyword: {keyword}')
    for result in results:
        print(f'  {result}')
    print()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Given dictionary structure
input_dict = {
    'F': [
        {'F[Hepatitis_B,(3, 7)]': 1.0},
        {'F[Hepatitis_B,(5, 5)]': 0.0},
        {'F[Mumps,(1, 1)]': 1.0},
        {'F[Polio,(2, 4)]': 0.0},
        {'F[Tetanus,(3, 3)]': 1.0},
    ],
}

# Extract data from the first DataFrame
group_data_list = input_dict['F']
group_data = []

for dict_item in group_data_list:
    # Extracting the key from the dictionary
    key = list(dict_item.keys())[0]

    # Splitting the key into three values
    name = key.split('[', 1)[1].split(',', 1)[0]
    start, finish = eval(key.split('(', 1)[1].split(')', 1)[0])
    value = dict_item[key]

    # Append data to the current group
    group_data.append({'Name': name, 'Start': start, 'Finish': finish, 'Value': value})

# Create a DataFrame for the current group data
df = pd.DataFrame(group_data)

# Group by 'Name'
grouped_df = df.groupby('Name').agg({'Start': 'min', 'Finish': 'max', 'Value': 'sum'}).reset_index()

# Sort DataFrame by 'Start'
grouped_df = grouped_df.sort_values('Start')

# Plotting
fig, ax = plt.subplots()

bar_width = 0.35

for index, row in grouped_df.iterrows():
    # Plotting each section of the bar individually
    for i in range(int(row['Start']), int(row['Finish']) + 1):
        color = 'green' if row['Value'] == 1 else 'red'
        ax.barh(row['Name'], 1, left=i, height=bar_width, color=color, edgecolor='black')

ax.set_xlabel('Start and End')
ax.set_title('Horizontal Bar Plot with Portions Corresponding to Each Bin')

# Set x-axis limit to 10
plt.xlim(0, 10)
# Set x-axis ticks at each 1 unit increase
plt.xticks(np.arange(0, 11, 1))

plt.show()
```
