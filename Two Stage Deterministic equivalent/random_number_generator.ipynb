```julia
using Distributions

# Function to calculate NORM.INV equivalent
function norm_inv(p, μ=0, σ=1)
    dist = Normal(μ, σ)
    return quantile(dist, p)
end

# Function to enforce bounds on the quantile value
function enforce_bounds(value, lower_bound, upper_bound)
    if value < lower_bound
        return lower_bound
    elseif value > upper_bound
        return upper_bound
    else
        return value
    end
end


```

```julia
# Example usage
probability = 0.95
mean = 457986356.1
std_dev = 2.90E+08
lower_bound = 247261081.2
upper_bound = 668711631

result = norm_inv(rand(), mean, std_dev)
bounded_result = enforce_bounds(result, lower_bound, upper_bound)

println("The computed inverse of the normal distribution is $result")
println("The bounded result is $bounded_result")
```
