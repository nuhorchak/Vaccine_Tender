```python
import pandas as pd
```

```python
# Path to your Excel file
file_path = 'random_normal_forecast_data.xlsx'

# Load the Excel file
xl = pd.ExcelFile(file_path)

# Read all sheets into a list of dataframes, assuming the first row and column have labels
dfs = [xl.parse(sheet_name, index_col=0) for sheet_name in xl.sheet_names]

# Use the first DataFrame to establish the structure of the result DataFrame
result_df = pd.DataFrame(index=dfs[0].index, columns=dfs[0].columns)

# Calculate the average of corresponding cells
for column in result_df.columns:
    for index in result_df.index:
        # Extract the cell values across all sheets for the current cell
        cell_values = [df.at[index, column] for df in dfs]
        # Compute the average and assign it to the corresponding cell in the result DataFrame
        result_df.at[index, column] = sum(cell_values) / len(cell_values)

# Save the result to a new Excel file
result_df.to_excel('averaged_sheets.xlsx')

```
