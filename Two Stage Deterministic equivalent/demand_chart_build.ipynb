```python
import pandas as pd

# Load the Excel file
# file_path = '../production_capacity_updated.xlsx'
file_path = '../MVP_random_normal_forecast_data.xlsx'



# Read the 'Medium_Capacity' sheet
data_medium_capacity = pd.read_excel(file_path, sheet_name='Medium_Demand')  #Medium_Capacity
data_medium_capacity.head()
```

```python
import matplotlib.pyplot as plt

# Assuming the years are sequential and simply labeled as 1 through 10
years = list(range(1, 11))

# Set up the plot
plt.figure(figsize=(12, 8))

# Loop through each row to plot the data
for index, row in data_medium_capacity.iterrows():
    plt.plot(years, row[1:], label=row['Unnamed: 0'])  # Skip the first column which is the name

# Adding titles and labels
plt.title('Antigen Demand', fontsize=20)
plt.xlabel('Year')
plt.ylabel('Demand')
plt.xticks(years, [str(year) for year in range(1, 1 + len(years))])  # Assuming the years start from 2014

# Adding legend to the left
plt.legend(loc='center left', bbox_to_anchor=(1, 0.5))

# Show plot
plt.tight_layout()
# plt.show()

plt.savefig('demand.pdf', bbox_inches='tight', pad_inches=0.1, dpi=600)
```
