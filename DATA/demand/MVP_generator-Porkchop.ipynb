```python
import pandas as pd
import numpy as np
```

```python
def calculate_average(dataframes_dict, num_to_consider=None):
    # Get the list of dataframes
    dataframes_list = list(dataframes_dict.values())
    
    # If num_to_consider is specified, use only that many dataframes
    if num_to_consider is not None:
        dataframes_list = dataframes_list[:num_to_consider]
    
    # Ensure all dataframes have the same shape
    first_shape = dataframes_list[0].shape
    if not all(df.shape == first_shape for df in dataframes_list):
        raise ValueError("All dataframes must have the same shape")
    
    # Initialize an empty dataframe to store the average values
    average_df = pd.DataFrame(np.zeros(first_shape), columns=dataframes_list[0].columns)
    
    # Calculate the average for each cell
    for df in dataframes_list:
        # Convert all data to numeric, filling non-numeric values with zeros
        numeric_df = df.apply(pd.to_numeric, errors='coerce').fillna(0)
        average_df += numeric_df
    
    # Divide by the number of dataframes to get the average
    average_df /= len(dataframes_list)
    
    return average_df

# Example usage
file_path = 'random_normal_forecast_data.xlsx'
excel_file = pd.ExcelFile(file_path)
dataframes_dict = {sheet_name: pd.read_excel(file_path, sheet_name=sheet_name) for sheet_name in excel_file.sheet_names}

# Calculate average using the first 3 dataframes
average_df = calculate_average(dataframes_dict, num_to_consider=30)

# Set the row labels
row_labels = [
    "Measles", "Mumps", "Rubella", "Diphtheria", "Tetanus",
    "Pertussis", "Hepatitis_B", "Hib", "Polio", "HPV",
    "Rotavirus", "PCV"
]
average_df.insert(0, 'Vaccine', row_labels)
average_df = average_df.drop(columns=['Unnamed: 0'])

# print("Average DataFrame:")
print(average_df)

average_df.to_excel('MVP30_random_normal_forecast_data.xlsx', index=False)
```
