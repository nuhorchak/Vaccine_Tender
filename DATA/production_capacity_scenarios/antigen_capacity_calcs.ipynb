```python
import pandas as pd
import numpy as np
```

```python
# Load the data from Sheet1
file_path = 'ARCHIVE/producer_portfolio_capacity.xlsx'
xls = pd.ExcelFile(file_path)
sheet1_df = pd.read_excel(xls, sheet_name='Sheet1')

# Display the first few rows of Sheet1 to understand its structure
sheet1_df.head()
```

```python
# Define the dictionary with mappings
A_v = {
    "Measles": ["Measles"], "MR": ["Measles", "Rubella"], "MMR": ["Measles", "Mumps", "Rubella"], 
    "TT": ["Tetanus"], "HepB": ["Hepatitis_B"], "Hib": ["Hib"], "IPV": ["Polio"], 
    "OPV": ["Polio"], "DT": ["Diphtheria", "Tetanus"], "Td": ["Diphtheria", "Tetanus"], 
    "DTwP": ["Diphtheria", "Tetanus", "Pertussis"], "DTwP-Hib": ["Diphtheria", "Tetanus", "Pertussis", "Hib"], 
    "Penta": ["Diphtheria", "Tetanus", "Pertussis", "Hepatitis_B", "Hib"], 
    "Hexa": ["Diphtheria", "Tetanus", "Pertussis", "Hepatitis_B", "Hib", "Polio"], 
    "HPV": ["HPV"], "Rota": ["Rotavirus"], "PCV": ["PCV"]
}

# Flatten the dictionary to create a reverse mapping
reverse_mapping = {}
for key, values in A_v.items():
    for value in values:
        if value not in reverse_mapping:
            reverse_mapping[value] = []
        reverse_mapping[value].append(key)

reverse_mapping

```

```python
# Initialize a dictionary to store the aggregated adjusted capacity
aggregated_capacity = {vaccine: 0 for vaccine in reverse_mapping.keys()}

# Iterate through each row in the dataframe and update the aggregated adjusted capacity
for index, row in sheet1_df.iterrows():
    vaccine = row['Vaccine']
    # print(vaccine)
    if vaccine in A_v:
        for disease in A_v[vaccine]:
            # print(disease)
            aggregated_capacity[disease] += row['Adjusted Capacity']
            # print(row['Adjusted Capacity'])

aggregated_capacity

```
