```python
import pandas as pd

# Load the workbook
file_path = 'wuenic2022rev_web-update.xlsx'
xls = pd.ExcelFile(file_path)

# Provided GAVI countries list
gavi_countries = [
    "Afghanistan", "Burkina Faso", "Burundi", "Central African Republic", "Chad", 
    "Democratic Republic of the Congo", "Democratic People's Republic of Korea", 
    "Eritrea", "Ethiopia", "Gambia", "Guinea-Bissau", "Liberia", "Madagascar", 
    "Malawi", "Mali", "Mozambique", "Niger", "Rwanda", "Sierra Leone", "Somalia", 
    "South Sudan", "Sudan", "Syrian Arab Republic", "Togo", "Uganda", "Yemen", 
    "Benin", "Cambodia", "Cameroon", "Comoros", "Guinea", "Haiti", "Kyrgyzstan", 
    "Lesotho", "Mauritania", "Myanmar", "Nepal", "Pakistan", "Senegal", "Tajikistan", 
    "UR Tanzania", "Zambia", "Zimbabwe", "Bangladesh", "Congo", "Côte d'Ivoire", 
    "Djibouti", "Ghana", "Kenya", "Lao People’s Democratic Republic", "Nigeria", 
    "Papua New Guinea", "Sao Tome and Principe", "Solomon Islands"
]

# Initialize a dictionary to hold the average coverage for each sheet
average_coverage = {}

for sheet_name in xls.sheet_names:
    # Skip the 'regional_global' sheet as it has a different structure
    if sheet_name == 'regional_global':
        continue

    df = pd.read_excel(xls, sheet_name=sheet_name)
    
    # Filter out countries not in the GAVI list
    df_filtered = df[df['country'].isin(gavi_countries)]
    
    # Calculate the average coverage for the year 2022
    avg_coverage = df_filtered['2022'].mean()
    
    # Store the average coverage in the dictionary
    average_coverage[sheet_name] = avg_coverage

# Convert the dictionary to a DataFrame for reporting
average_coverage_df = pd.DataFrame(list(average_coverage.items()), columns=['Antigen', 'Average Coverage 2022'])

average_coverage_df

```
